```python
# Enter Telegram credentials securely (nothing is printed back)
import getpass, os

BOT_TOKEN = getpass.getpass('BOT_TOKEN: ').strip()
TELEGRAM_API_ID = getpass.getpass('TELEGRAM_API_ID: ').strip()
TELEGRAM_API_HASH = getpass.getpass('TELEGRAM_API_HASH: ').strip()

if not BOT_TOKEN or not TELEGRAM_API_ID or not TELEGRAM_API_HASH:
    raise ValueError('BOT_TOKEN, TELEGRAM_API_ID and TELEGRAM_API_HASH are all required.')

os.environ['BOT_TOKEN'] = BOT_TOKEN
os.environ['TELEGRAM_API_ID'] = TELEGRAM_API_ID
os.environ['TELEGRAM_API_HASH'] = TELEGRAM_API_HASH
os.environ['DOWNLOAD_DIR'] = './downloads'
os.environ['TEMP_DIR'] = './tmp'
os.environ['MAX_TELEGRAM_FILE_SIZE'] = str(2 * 1024**3)
os.environ['PYTHONUNBUFFERED'] = '1'

# Do not print the secret values.
del BOT_TOKEN, TELEGRAM_API_ID, TELEGRAM_API_HASH
print('Telegram credentials loaded into this Colab runtime.')
```
